```python
#Perfilado de tipos: por cada columna se infiere si puede ser int, double, date o booleano
#y se listan los valores que impiden la conversion (incidencias)
DATASETS = {
    "divipola": "../data/raw/divipola.csv",
    "eva": "../data/raw/eva.csv",
    "agrosavia": "../data/raw/agrosavia.csv",
}
MARCADORES_NULO = {"ND", "N.D.", "N.D", "NA", "N/A", "#N/A", "NAN", "NULL", "NONE", "-", "--",
                   "NO INDICA", "NO REPORTA", "SIN DATO", "SIN INFORMACION", "SIN INFORMACIÓN", "S.I."}
VERDADEROS = {"TRUE", "VERDADERO", "SI", "SÍ", "S", "YES", "Y", "T"}
FALSOS = {"FALSE", "FALSO", "NO", "N", "F"}
FORMATOS_FECHA = ["%d/%m/%Y", "%Y-%m-%d", "%d-%m-%Y", "%Y/%m/%d", "%m/%d/%Y",
                  "%d/%m/%Y %H:%M:%S", "%Y-%m-%d %H:%M:%S"]
FECHA_MIN, FECHA_MAX = pd.Timestamp("1950-01-01"), pd.Timestamp.today()
UMBRAL_TIPO = 0.8  #% minimo de valores reconocibles para asumir que la columna NO es str
ACCION = {
    "marcador_nulo":    "reemplazar por NaN/NaT antes de convertir",
    "limite_deteccion": "valor censurado (<x / >x): decidir si usar x, x/2 o NaN",
    "coma_decimal":     "cambiar ',' por '.' antes de convertir",
    "cero_izquierda":   "es un codigo: mantener como str",
    "fuera_de_rango":   "fecha improbable: revisar",
    "otro":             "revisar manualmente",
}
#Incidencias que no impiden reconocer el tipo porque se pueden corregir
RECUPERABLES = {"ok", "cero_izquierda", "coma_decimal", "limite_deteccion", "fuera_de_rango"}


#Los clasificadores reciben los valores unicos de la columna y marcan cada uno como
#"ok", "vacio" o el tipo de incidencia que impide convertirlo
def marcar_nulos(cat, v):
    cat[v.str.upper().isin(MARCADORES_NULO)] = "marcador_nulo"
    cat[v.eq("")] = "vacio"
    return cat

def clasificar_numero(v):
    cat = pd.Series("otro", index=v.index)
    cat[pd.to_numeric(v, errors="coerce").notna()] = "ok"
    cat[v.str.fullmatch(r"0\d+")] = "cero_izquierda"
    cat[v.str.fullmatch(r"-?\d+(?:\.\d{3})*,\d+")] = "coma_decimal"
    cat[v.str.fullmatch(r"[<>]=?\s*-?\d+(?:[.,]\d+)?")] = "limite_deteccion"
    return marcar_nulos(cat, v), None

def clasificar_fecha(v):
    #Se usa el formato que logra convertir mas valores
    mejor_fmt, fechas = max(((fmt, pd.to_datetime(v, format=fmt, errors="coerce")) for fmt in FORMATOS_FECHA),
                            key=lambda par: par[1].notna().sum())
    cat = pd.Series("otro", index=v.index)
    cat[fechas.notna()] = "ok"
    cat[fechas.lt(FECHA_MIN) | fechas.gt(FECHA_MAX)] = "fuera_de_rango"
    return marcar_nulos(cat, v), mejor_fmt

def clasificar_booleano(v):
    cat = pd.Series("otro", index=v.index)
    cat[v.str.upper().isin(VERDADEROS | FALSOS)] = "ok"
    return marcar_nulos(cat, v), None

def perfilar_columna(serie):
    conteo = serie.str.strip().value_counts()  #valor unico -> frecuencia
    v = pd.Series(conteo.index, index=conteo.index)
    #Se prueba cada tipo y se queda el de mayor % de valores reconocibles
    mejor = None
    for tipo, clasificador in [("numero", clasificar_numero), ("fecha", clasificar_fecha), ("booleano", clasificar_booleano)]:
        if tipo == "fecha" and not v.str.contains(r"\d[/-]\d").any():
            continue
        cat, fmt = clasificador(v)
        con_dato = conteo[~cat.isin(["vacio", "marcador_nulo"])].sum()
        ratio = conteo[cat.isin(RECUPERABLES)].sum() / con_dato if con_dato else 0
        if mejor is None or ratio > mejor[1]:
            mejor = (tipo, ratio, cat, fmt)
    tipo, ratio, cat, fmt = mejor
    if ratio < UMBRAL_TIPO:
        return "str", None, conteo
    if tipo == "numero":
        if cat.eq("cero_izquierda").any():
            tipo = "str (codigo)"
        else:
            tipo = "int" if v[cat.isin(RECUPERABLES)].str.fullmatch(r"-?\d+").all() else "double"
    elif tipo == "fecha":
        tipo = f"date ({fmt})"
    return tipo, cat, conteo

def perfilar_dataset(nombre, ruta, max_ejemplos=5):
    #Todo se lee como texto y sin convertir vacios a NaN para ver el valor original
    df = pd.read_csv(ruta, dtype=str, keep_default_na=False)
    filas_resumen, filas_detalle = [], []
    for columna in df.columns:
        tipo, cat, conteo = perfilar_columna(df[columna])
        fila = {"dataset": nombre, "columna": columna, "tipo_sugerido": tipo,
                "filas": len(df), "vacios": int(conteo.get("", 0)), "convertibles": None, "incidencias": 0}
        if cat is not None:
            fila["convertibles"] = int(conteo[cat.eq("ok")].sum())
            incid = cat[~cat.isin(["ok", "vacio"])]
            for tipo_inc, valores in incid.groupby(incid):
                frec = conteo[valores.index].sort_values(ascending=False)
                fila[tipo_inc] = int(frec.sum())
                fila["incidencias"] += int(frec.sum())
                filas_detalle.append({
                    "dataset": nombre, "columna": columna, "tipo_sugerido": tipo, "incidencia": tipo_inc,
                    "filas_afectadas": int(frec.sum()), "valores_distintos": len(frec),
                    "ejemplos": ", ".join(f"'{k}' ({n})" for k, n in frec.head(max_ejemplos).items()),
                    "accion": ACCION[tipo_inc],
                })
        filas_resumen.append(fila)
    return filas_resumen, filas_detalle

resumen, detalle = [], []
for nombre, ruta in DATASETS.items():
    filas_resumen, filas_detalle = perfilar_dataset(nombre, ruta)
    resumen += filas_resumen
    detalle += filas_detalle

resumen = pd.DataFrame(resumen)
tipos_inc = [t for t in ACCION if t in resumen.columns]
resumen[tipos_inc] = resumen[tipos_inc].fillna(0).astype(int)
resumen["convertibles"] = resumen["convertibles"].astype("Int64")
resumen["pct_incidencias"] = (resumen["incidencias"] / resumen["filas"] * 100).round(2)
resumen = resumen[["dataset", "columna", "tipo_sugerido", "filas", "vacios", "convertibles",
                   "incidencias", "pct_incidencias"] + tipos_inc]
detalle = pd.DataFrame(detalle)
```
