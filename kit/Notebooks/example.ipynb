```python
import pandas as pd

produccion = pd.DataFrame({
    "codigo_municipio": [15001, 15238, 15516],
    "cultivo": ["Papa", "Cebolla", "Arveja"],
    "toneladas": [1200, 850, 430],
})

display(produccion)
```

```python
municipios = pd.DataFrame({
    "codigo_municipio": [15001, 15238, 15759],
    "nombre_municipio": ["Tunja", "Duitama", "Sogamoso"],
})

display(municipios)
```

```python
inner = pd.merge(produccion, municipios, on="codigo_municipio", how="inner")

display(inner)
```

```python
left = pd.merge(produccion, municipios, on="codigo_municipio", how="left")

display(left)
```

```python
right = pd.merge(produccion, municipios, on="codigo_municipio", how="right")

display(right)
```
